```python
import pandas as pd

# List of your 6 spreadsheets
files = [
    "Results Band 1 wo League Adjust 24-25 and 25-26 ALL.xlsx",
    "Results Band 2 wo League Adjust 24-25 and 25-26 ALL.xlsx",
    "Results Band 3 wo League Adjust 24-25 and 25-26 ALL.xlsx",
    "Results Band 4 wo League Adjust 24-25 and 25-26 ALL.xlsx",
    "Results Band 5 wo League Adjust 24-25 and 25-26 ALL.xlsx",
    "Results Band 6 wo League Adjust 24-25 and 25-26 ALL.xlsx"
]

# Name of the output file
output_file = "combined_band_sheets_09-12-25.xlsx"

# Create a writer object
with pd.ExcelWriter(output_file, engine="openpyxl") as writer:
    for i, file in enumerate(files, start=1):
        # Read each spreadsheet into a DataFrame
        df = pd.read_excel(file)  # use pd.read_csv(file) if it's a CSV
        # Save to a separate sheet
        sheet_name = f"Sheet{i}"  # you can customize with file names
        df.to_excel(writer, sheet_name=sheet_name, index=False)

print(f"✅ Combined file saved as {output_file}")
```
